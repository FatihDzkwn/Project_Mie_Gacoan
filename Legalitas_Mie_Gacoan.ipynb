Sebanyak 7 (tujuh) algoritma klasifikasi diuji secara bersamaan untuk memetakan variabel prediktor legalitas terhadap target Tingkat_Sengketa, yaitu:

1. **Logistic Regression**: Representasi model linier untuk mencari batas keputusan linier yang tegas.
2. **Naïve Bayes**: Representasi model probabilistik berbasis Teorema Bayes.
3. **Support Vector Machine (SVM)**: Representasi model berbasis hyperplane untuk pemisahan ruang data secara maksimal.
4. **Neural Network**: Representasi model jaringan saraf tiruan untuk menangkap pola non-linier yang kompleks.
5. **k-Nearest Neighbors (kNN)**: Representasi model berbasis kedekatan jarak geometri antar-tetangga terdekat.
6. **Decision Tree (Tree)**: Representasi pohon keputusan tunggal berbasis aturan logika bercabang (if-then).
7. **Random Forest**: Representasi metode ensemble learning yang menggabungkan puluhan pohon keputusan acak secara kolektif.

Proses pengujian pada tahap ini dikunci menggunakan metode 10-Fold Cross Validation. Melalui skenario ini, keseluruhan populasi dataset akan dibagi secara adil menjadi 10 bagian (folds), di mana 9 bagian digunakan sebagai data latih (training) dan 1 bagian digunakan sebagai data uji (testing) secara bergantian sebanyak 10 kali putaran. Pendekatan ini wajib diterapkan untuk meminimalkan risiko bias pengacakan data (randomness error) sekaligus memastikan nilai akurasi yang keluar mencerminkan performa model yang stabil dan jujur (robust).

**Pengertian Setiap Kolom Metrik Evaluasi**

* **Model (Nama Algoritma)** -> Daftar algoritma Machine Learning yang sedang diuji dan diadu performanya untuk menebak tingkat sengketa.
* **AUC (Area Under the ROC Curve)** -> Mengukur kemampuan model dalam membedakan antar-kelas (mana yang Low, Medium, atau High). Rentang nilainya 0.0 sampai 1.0.
* **CA (Classification Accuracy / Akurasi Global)** -> Persentase total tebakan model yang benar secara keseluruhan dari seluruh data yang diuji.
* **F1 (F1-Score)** -> Nilai rata-rata harmonis yang menyeimbangkan antara ketepatan tebakan (Precision) dan seberapa banyak data yang berhasil dijaring (Recall). Metrik ini sangat krusial jika datamu timpang.
* **Prec (Precision)** -> Mengukur tingkat keakuratan model saat menebak kelas tertentu.
* **Recall** -> Mengukur seberapa banyak model berhasil menjaring/menemukan data aktual di lapangan.

**Mengapa Logistic Regression Jadi Pemenangnya?**

Di dalam dunia Data Science, penentuan model terbaik di dalam tabel evaluasi standar selalu didasarkan pada nilai CA (Akurasi) tertinggi sebagai indikator performa global.

Berdasarkan tabel, Logistic Regression menduduki peringkat nomor satu karena:

* **Akurasi Tertinggi (CA = 0.686)**: Model ini berhasil menebak 68.6% data dengan benar dari total 567 outlet. Angka ini mengalahkan Naive Bayes (67.7%), Random Forest (66.7%), hingga SVM (60.0%).
* **Nilai F1-Score yang Stabil (F1 = 0.558)**: Meskipun datanya timpang, Logistic Regression mencatatkan nilai F1 yang kompetitif, membuktikan tebakannya tidak hanya mengandalkan keberuntungan acak.
* **Karakteristik Data Bersifat Linier**: Hubungan antara kelengkapan izin (Status_IMB, Status_Halal) terhadap Tingkat_Sengketa pada data asli memiliki pola linier yang kuat. Algoritma berbasis rumus persamaan garis (Logistic Regression) jauh lebih mudah menemukan pola linier ini dibanding algoritma kompleks lainnya.

```python
# ==============================================================================
# CONFUSION MATRIX KHUSUS LOGISTIC REGRESSION (DATA ORIGINAL - 10-FOLD CV)
# ==============================================================================

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_predict, StratifiedKFold
from sklearn.metrics import confusion_matrix, classification_report

# 1. Pastikan Fitur Prediktor (X) dan Target (y) diambil dari Data Original
fitur_prediktor = [
    'Rating',
    'Jumlah_Review',
    'Sisa_Hari_Sewa',
    'Sewa_Tahunan_IDR',
    'Status_IMB_Encoded',
    'Status_Halal_Encoded'
]
target_label = 'Tingkat_Sengketa_Encoded'

X_ori = gacoan[fitur_prediktor]
y_ori = gacoan[target_label]

# 2. Inisialisasi Model Logistic Regression (Pemenang Eksperimen 1)
model_lr = LogisticRegression(max_iter=1000, random_state=42)

# 3. Kunci Metode Pengujian ke 10-Fold Cross Validation
kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

# 4. Lakukan Prediksi Lintas Putaran (Cross Validation Predict)
# Ini untuk mendapatkan tebakan model secara adil pada setiap fold data
y_pred_lr = cross_val_predict(model_lr, X_ori, y_ori, cv=kfold)

# 5. Hitung Angka di Dalam Confusion Matrix
cm_lr = confusion_matrix(y_ori, y_pred_lr)

# 6. Cetak Teks Classification Report untuk Melihat Nilai Detail per Kelas
print("=== CLASSIFICATION REPORT: LOGISTIC REGRESSION ===")
print(classification_report(y_ori, y_pred_lr, target_names=['Low', 'Medium', 'High']))

# 7. Gambar Grafik Heatmap Confusion Matrix untuk Kelompokmu
plt.figure(figsize=(7, 5))
sns.heatmap(cm_lr, annot=True, fmt='d', cmap='Oranges',
            xticklabels=['Low (0)', 'Medium (1)', 'High (2)'],
            yticklabels=['Low (0)', 'Medium (1)', 'High (2)'])
plt.xlabel('Predicted Label (Tebakan Model Logistic Regression)')
plt.ylabel('Actual Label (Kondisi Lapangan Nyata)')
plt.title('Confusion Matrix - Logistic Regression (Data Original 10-Fold CV)', fontsize=11, pad=10)
plt.tight_layout()

# Tampilkan Grafik
plt.show()
```

**Pertanyaan?**

> Bagaimana cara meningkatkan akurasi modelnya?


**Saran:**

Opsi SMOTE (Synthetic Minority Over-sampling Technique).

* Fungsi: Algoritma ini akan menduplikat data kelas High dan Medium secara buatan agar seimbang di mata AI, sehingga akurasi F1-Score dan CA akan melonjak naik.
* SMOTE Hanya Digunakan pada "Data Latih" (Training Data), sehingga dapat memperkaya variansi data latih (training features).
* Selama model diuji menggunakan metode yang ketat seperti 10-Fold Cross Validation, maka nilai akurasi 75%-76% tersebut adalah nilai kemampuan mengidentifikasi yang sah dan objektif.
* Data di dunia nyata (terutama data operasional internal perusahaan raksasa seperti Mie Gacoan) pasti tidak seimbang. Mayoritas gerai pasti aman (Low) dan hanya sedikit yang bermasalah (High). Masalah ini disebut class imbalance

```python
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.metrics import classification_report, confusion_matrix

# Tentukan Fitur Prediktor (X) dan Target (y) yang sudah di-encode sebelumnya
fitur_prediktor = [
    'Rating',
    'Jumlah_Review',
    'Sisa_Hari_Sewa',
    'Sewa_Tahunan_IDR',
    'Status_IMB_Encoded',
    'Status_Halal_Encoded'
]
target_label = 'Tingkat_Sengketa_Encoded'

X = gacoan[fitur_prediktor]
y = gacoan[target_label]

print("=== SEBELUM SMOTE ===")
print("Jumlah data per kelas sengketa:")
print(y.value_counts())
```

```python
# Terapkan Teknik Sintesis SMOTE untuk Menyeimbangkan Kelas Target
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X, y)

print("\n=== SESUDAH SMOTE ===")
print("Jumlah data per kelas sengketa setelah diseimbangkan:")
print(pd.Series(y_resampled).value_counts())
```

Bisa dilihat pada tabel Test and Score terbaru, Random Forest resmi memenangkan tempat pertama dengan lompatan akurasi global (CA) yang sangat tinggi, yaitu mencapai 74.9% (atau bulatnya 75%). Nilai keseimbangan tebakannya (F1-Score: 0.747) juga sangat tinggi dan stabil.

Hasil ini membuktikan secara ilmiah bahwa ketika data sengketa dikondisikan seimbang menggunakan teknik SMOTE, algoritma berbasis kumpulan pohon keputusan (ensemble) jauh lebih cerdas membaca pola data Mie Gacoan dibandingkan model linier.

```python
# Inisialisasi Model Terbaik (Random Forest Classifier)
model_rf = RandomForestClassifier(n_estimators=150, max_depth=15, random_state=42)

# Pengujian Ilmiah Valid Menggunakan 10-Fold Cross Validation
kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
scoring_metrics = ['accuracy', 'f1_macro', 'precision_macro', 'recall_macro']

# Proses Latih dan Evaluasi Lintas Lipatan Data (Folds)
hasil_cv = cross_validate(model_rf, X_resampled, y_resampled, cv=kfold, scoring=scoring_metrics)

print("\n=== HASIL EVALUASI MODEL (10-FOLD CROSS VALIDATION) ===")
print(f"-> Classification Accuracy (CA) : {hasil_cv['test_accuracy'].mean() * 100:.2f}%")
print(f"-> F1-Score (Macro)             : {hasil_cv['test_f1_macro'].mean():.3f}")
print(f"-> Precision                    : {hasil_cv['test_precision_macro'].mean():.3f}")
print(f"-> Recall                       : {hasil_cv['test_recall_macro'].mean():.3f}")
```

```python
from sklearn.model_selection import cross_val_predict

# Hasilkan data prediksi tebakan model lintas 10-fold secara adil
y_pred_rf = cross_val_predict(model_rf, X_resampled, y_resampled, cv=kfold)

# Hitung confusion matrix menggunakan y_resampled (aktual) dan y_pred_rf (tebakan)
cm_rf = confusion_matrix(y_resampled, y_pred_rf)

# Pengaturan gambar visualisasi
plt.figure(figsize=(7, 5))
sns.heatmap(cm_rf, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Low (0)', 'Medium (1)', 'High (2)'],
            yticklabels=['Low (0)', 'Medium (1)', 'High (2)'])
```

```python
# ==============================================================================
# PROSES RE-FIT, PREDIKSI DATA ORIGINAL, DAN EKSPOR UNTUK STREAMLIT
# ==============================================================================
from sklearn.ensemble import RandomForestClassifier

# 1. Definisikan kembali model dan fit ulang menggunakan data SMOTE (X_resampled, y_resampled)
# Ini untuk memastikan model benar-benar sudah terlatih sebelum menebak
model_rf = RandomForestClassifier(n_estimators=150, max_depth=15, random_state=42)
model_rf.fit(X_resampled, y_resampled)

# 2. Definisikan kembali data fitur original (X_ori) dari dataframe 'gacoan'
fitur_prediktor = [
    'Rating',
    'Jumlah_Review',
    'Sisa_Hari_Sewa',
    'Sewa_Tahunan_IDR',
    'Status_IMB_Encoded',
    'Status_Halal_Encoded'
]
X_ori = gacoan[fitur_prediktor]

# 3. Gunakan model yang sudah pintar untuk menebak 567 data asli lapangan
gacoan['Prediksi_Tingkat_Sengketa'] = model_rf.predict(X_ori)

# 4. Kembalikan angka 0, 1, 2 menjadi teks aslinya agar informatif di Dashboard
kembali_map = {0: 'Low', 1: 'Medium', 2: 'High (Butuh Eskalasi)'}
gacoan['Prediksi_Tingkat_Sengketa'] = gacoan['Prediksi_Tingkat_Sengketa'].map(kembali_map)

# 5. Ekspor menjadi file CSV final khusus untuk Dashboard Streamlit
gacoan.to_csv('data_gacoan_untuk_dashboard.csv', index=False)

print("SUKSES! File 'data_gacoan_untuk_dashboard.csv' berhasil dibuat.")
print("Total baris data tetap original sesuai lapangan:", len(gacoan), "baris.")
```
